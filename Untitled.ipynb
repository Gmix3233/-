```python
import math

def classify_number(n):
    """
    Определяет чётность (для целых) и классифицирует число по видам.
    """
    # Блок 1: Определение чётности
    if isinstance(n, int) or (isinstance(n, float) and n.is_integer()):
        is_even = int(n) % 2 == 0
        parity_str = "чётное" if is_even else "нечётное"
        print(f"Число {n}: {parity_str}.")
    else:
        print(f"Число {n}: не является целым, понятие чётности к нему не применяется.")

    # Блок 2: Классификация по видам чисел
    is_integer = isinstance(n, int) or (isinstance(n, float) and n.is_integer())
    is_real = isinstance(n, (int, float))
    is_rational = isinstance(n, (int, float))
    is_irrational = False

    # Проверка на популярные иррациональные константы
    if isinstance(n, float):
        if math.isclose(n, math.pi) or math.isclose(n, math.e):
            is_irrational = True
            is_rational = False

    # Проверка на корни (если корень не целый, число иррационально)
    if isinstance(n, float) and n > 0 and not n.is_integer():
        root = math.sqrt(n)
        if not root.is_integer():
            is_irrational = True
            is_rational = False

    is_natural = is_integer and n > 0
    is_whole_non_negative = is_integer and n >= 0

    # Вывод классификации
    print("Классификация по видам чисел:")
    print(f"  - Действительное: {'Да' if is_real else 'Нет'}")
    print(f"  - Рациональное: {'Да' if is_rational else 'Нет'}")
    print(f"  - Иррациональное: {'Да' if is_irrational else 'Нет'}")
    print(f"  - Целое: {'Да' if is_integer else 'Нет'}")
    print(f"  - Натуральное (1, 2, 3...): {'Да' if is_natural else 'Нет'}")
    print(f"  - Целое неотрицательное (0, 1, 2...): {'Да' if is_whole_non_negative else 'Нет'}")
    print("=" * 30)


# Блок 3: Тестирование
if __name__ == "__main__":
    # Вы можете менять значения в этом списке для проверки своих чисел
    test_cases = [
        24,              # целое положительное
        -15,             # целое отрицательное
        0,               # ноль
        8.0,             # float, который является целым
        3.14,            # обычная дробь
        math.sqrt(49),   # корень из квадрата (даст 7.0)
        math.sqrt(20),   # иррациональный корень
        math.pi          # число Пи
    ]

    for number in test_cases:
        classify_number(number)
    
```
